```python
import pandas as pd
```

```python
d=pd.read_csv('clean_final_data.csv')
d
```

```python
Total_Order=len(d)
print(Total_Order)
```

```python
Unique_Customer=d['CustomerID'].nunique()
print(Unique_Customer)
```

```python
d['OrderDate']=pd.to_datetime(d['OrderDate'])
```

```python
d['SignupDate']=pd.to_datetime(d['SignupDate'])
```

```python
d['OrderDate'].dtype
```

```python
d['SignupDate'].dtype
```

```python
Total_Sales=d['Sales'].sum()
print(Total_Sales)
```

```python
Top_Sales_Products=d.groupby('ProductName')['Sales'].agg(['sum','count']).sort_values(by='sum',ascending=False)
print(Top_Sales_Product)
```

```python
Top_Sales_Product.head()
```

```python
Category_Performance=d.groupby('Category').agg(
    Top_Sales=('Sales','sum'),
    Avg_Discount=('Discount','mean'),
    Total_Quantity=('Quantity','sum')
)
print(Category_Performance)
```

```python
City_Performance=d.groupby('City').agg(
    Unique_Customer=('CustomerID','nunique'),
    Sales=('Sales','sum'),
    Total_Order=('Sales','count')
)
print(City_Performance)
```

```python
Payment_Method=d['PaymentMethod'].value_counts()
print(Payment_Method)
```

```python
Order_Status=d['Status'].value_counts()
print(Order_Status)
```

```python
Customer_Status=d['CustomerSegment'].value_counts()
print(Customer_Status)
```

```python
Customer_Order_Status=d.groupby('CustomerSegment').agg(
    Total_Sales=('Sales','sum'),
    Avg_Order_Value=('OrderValue','mean')
)
print(Customer_Order_Status)
```

```python
with pd.ExcelWriter('Final_ECommerce_Business_Report.xlsx') as writer:
    Category_Performance.to_excel(writer, sheet_name='Category Summary')
    City_Performance.to_excel(writer, sheet_name='City Performance')
    Top_Sales_Products.to_excel(writer, sheet_name='Top Products')
    Order_Status.to_excel(writer,sheet_name='Order Status')

print("Report successfully exported to Final_ECommerce_Business_Report.xlsx!")
```
